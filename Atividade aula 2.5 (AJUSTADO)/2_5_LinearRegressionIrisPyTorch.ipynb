# Regressão Linear com PyTorch

## Objetivo

Este exemplo utiliza regressão linear para estimar o comprimento das sépalas da íris a partir do seu comprimento das pétalas.
Utiliza-se
- o cálculo do gradiente automático do PyTorch, porém
- a função de perda e
- atualização dos parâmetros é feita de forma explícita.
- A rede é criada com uma camada nn.Linear()

![alt text](https://raw.githubusercontent.com/vcasadei/images/master/RegressaoLinear_Otimizacao.png)

## Importação dos pacotes

```python
%matplotlib inline
import matplotlib.pyplot as plt

import numpy as np
from sklearn.datasets import load_iris

import torch
from torch import nn, optim
from torch.autograd import Variable

torch.manual_seed(1234)
```

## Leitura dos dados

```python
iris = load_iris()
data = iris.data[iris.target==1,::2]  # comprimento das sépalas e pétalas, indices 0 e 2

x_train = data[:,0:1].astype(np.float32)
y_train = data[:,1:2].astype(np.float32)

n_samples = x_train.shape[0]
print('x_train.shape:',x_train.shape, x_train.dtype)
print('y_train.shape:',y_train.shape, y_train.dtype)

print('x_train[:5]:\n', x_train[:5])
print('y_train[:5]:\n', y_train[:5])
```

### Normalização dos dados

```python
x_train -= x_train.min()
x_train /= x_train.max()
y_train -= y_train.min()
y_train /= y_train.max()
```

```python
print('x_train[:5]:\n', x_train[:5])
print('y_train[:5]:\n', y_train[:5])
```

```python
x_train_bias = np.hstack([np.ones(shape=(n_samples,1)), x_train])
```

```python
x_train_bias
```

```python
print('x_train_bias.shape:',x_train_bias.shape, x_train_bias.dtype)
```

```python
x_train_bias = torch.FloatTensor(x_train_bias)
y_train      = torch.FloatTensor(y_train)
print(y_train)
```

## Model

![alt text](https://raw.githubusercontent.com/vcasadei/images/master/RegressaoLinearNeuronio.png)

```python
#Inicialização usando toch.nn.Linear

model = torch.nn.Linear(in_features=2, out_features=1, bias=False)
```

### Verificando a inicialização dos parâmetros

```python
model.weight.data = torch.zeros(1,2)
torch.nn.init.uniform(model.weight.data, -0.1, 0.1)
model.weight
```

### Testando o predict da rede

```python
xx = torch.ones((5,2))
xx
```

```python
model(Variable(xx))
```

## Treinamento

```python
num_epochs = 100
learning_rate = 0.5
for epoch in range(num_epochs):
    #inputs = Variable(x_train_bias)
    #target = Variable(y_train)
    inputs = x_train_bias
    target = y_train

    # forward - predict
    y_pred = model(inputs)

    #loss cálculo da função de perda
    loss = torch.mean((y_pred - target)**2)

    # backward propagation - cálculo gradient
    loss.backward()

    # gradiente descendente
    model.weight.data = model.weight.data - learning_rate * model.weight.grad.data

    # zeragem parâmetros próximo cálculo do gradiente
    model.weight.grad.data.zero_()

    # verbose
    if (epoch+1) % 20 == 0:
        print('Epoch[{}/{}], loss: {:.6f}'
              .format(epoch+1, num_epochs, loss.data))
```

## Avaliação

```python
y_pred = model(Variable(x_train_bias))
plt.plot(x_train, y_train.numpy(), 'ro', label='Original data')
plt.plot(x_train, y_pred.data.numpy(), 'kx-', label='Fitting Line')
plt.show()
```

### Acessando os parâmetros

```python
w_dic = model.state_dict()
w_dic
```

### Parâmetros ótimos, solução analítica

```python
x_bias = x_train_bias
y = y_train
x_bias_t = torch.t(x_bias)
w_opt = (torch.inverse(x_bias_t.mm(x_bias)).mm(x_bias_t)).mm(y)
print(torch.t(w_opt))
```

```python
# LISTAS PARA GUARDAR A EVOLUÇÃO DOS PESOS
w0_list = [0.1, 0.15, 0.2, 0.3]
w1_list = [0.1, 0.2, 0.18, 0.3]

#Inicialização usando torch.nn.Linear
model = torch.nn.Linear(in_features=2, out_features=1, bias=False)

#Inicialização manual dos parâmetros para que w0_list e w1_list sejam preenchidas antes do loop
w0_list.append(model.weight.data[0,0].item())
w1_list.append(model.weight.data[0,1].item())

num_epochs = 100
learning_rate = 0.5
for epoch in range(num_epochs):

    y_pred = model(x_train_bias)
    loss = torch.mean((y_pred - y_train)**2)

    loss.backward()

    model.weight.data -= learning_rate * model.weight.grad.data
    model.weight.grad.data.zero_()

    # guarda os pesos a cada época
    w0_list.append(model.weight.data[0,0].item())
    w1_list.append(model.weight.data[0,1].item())


    if (epoch+1) % 20 == 0:
        print(f"Epoch {epoch+1}/{num_epochs}, loss = {loss.item():.6f}")

# MSE final
y_pred = model(x_train_bias)
mse = torch.mean((y_pred - y_train)**2)
print("\nMSE final:", mse.item())

# solução analítica
x_bias = x_train_bias
y = y_train
x_bias_t = torch.t(x_bias)
w_opt = (torch.inverse(x_bias_t.mm(x_bias)).mm(x_bias_t)).mm(y)

w0_opt = w_opt[0].item()
w1_opt = w_opt[1].item()

print("\nPesos ótimos analíticos:", w0_opt, w1_opt)

# gráfico da reta ajustada
plt.figure()
plt.plot(x_train, y_train.numpy(), 'ro', label='Dados originais')
plt.plot(x_train, y_pred.detach().numpy(), 'kx-', label='Reta ajustada')
plt.legend()
plt.show()

# gráfico da trajetória dos pesos
plt.figure(figsize=(6,6))
plt.scatter(w0_list, w1_list, c=range(len(w0_list)), s=20)
plt.scatter(w0_opt, w1_opt, s=120, marker='X', color='red', label='Ótimo analítico')
plt.xlabel("w0 (bias)")
plt.ylabel("w1 (peso)")
plt.title("Caminho dos pesos durante o treino")
plt.colorbar(label="época")
plt.legend()
plt.show()
```

# Aprendizados

Percebi a diferença entre o bias e o peso, compreendi perfeitamente que no fundo de um modelo de regressão linear, é exatamente a aplicação matemática da função afim (função do primeiro grau) ou (função linear) f(x0) = a.x + b

No qual, trazendo para a realidade, é y = w.x + b

onde:

y -> a saída real do meu modelo;
y^ -> a saída esperada;
x -> entrada do omeu modelo
w -> o peso que ajusta a inclinação (o quão a variável de peso afeta ou não a entrada de dado da entrada sobre a saída y)
b -> bias, termo independente da função, juntamente com o peso ajuda a deslocar a reta

Se não mudarmos os pesos, a perda (loss) será exatamente a mesma, poisn não haverá reajustes no peso nem no bias, logo a reta da função afim, permanecerá a mesma

entendi que o ponto "ótimo" analítico da solução, é exatamente onde a reta corta o eixo verticar, ou seja, o eixo y, isto é aproximadamente 0,2613....

e que o,6864 é exatamente o que informa o quanto y sobe quando x sobe, isto é, o peso  w0 e w1 é diretamente proporcional ao y (saída do meu modelo), pos os pesos são positivos, tanto w0_list quanto w1_list são maiores que 0

Conclusão:

A solução analítica é a única linha possível que deixa o erro o menor possível, e por isso ela é a melhor linha que pode existir para aqueles dados.
